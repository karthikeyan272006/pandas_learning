Assign Function - to build transformation piplines

```python
import pandas as pd
```

```python
df= pd.read_csv("../data/cars.csv")
df.head()
```

```python
def convert_kg_to_ton(weight):
    return weight/1000
def convert_hp_to_kw(hp):
    return hp*0.745
```

```python
df['horsepower']=pd.to_numeric(df['horsepower'],errors='coerce')
```

```python
df.assign(horsepower = convert_hp_to_kw(df['horsepower']))
```
