Dataframe


```python
import pandas as pd
```

```python
apple =['Apple',60]
banana =['banana',50]
fruits = [apple,banana]
fruits
```

```python
df=pd.DataFrame(fruits)
df
```

```python
df.columns = ['fruits','price']
df
```

Dataframe using tuple

```python
apple =('Apple',60)
banana =('banana',50)
fruits = (apple,banana)

df = pd.DataFrame(fruits)
df.columns=['Fruits','price']
df
```

Dataframe using dictionary


```python
apple ={'name':'Apple','price':50}
banana ={'name':'banana','price':10}
mango ={'name':'mango','price':150}
fruits = (apple,banana,mango)
df =pd.DataFrame(fruits)
# df.columns =['fruits','price1']
df
```

```python
fruits ={
    'name':['apple','banana','mango'],
    'price':[50,10,100]
}
fruits
```

```python
df =pd.DataFrame(fruits)
df
```
