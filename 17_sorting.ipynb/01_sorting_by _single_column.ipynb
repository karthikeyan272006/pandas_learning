sorting data by single columns

```python
import pandas as pd
```

```python
df =pd.read_csv("../data/cars.csv")
df.head(5)
```

```python
df.sort_values(by = 'car name' ).head(10)
```

```python
df.sort_values(by = 'car name',ascending=False ).head(10)
```

```python
df.sort_values(by = 'mpg')
```

2. sorting rows by all columns

```python
df.sort_values(by = ['mpg','displacement'],ascending=[True,False])
```

3.prioriting null values

```python
df.sort_values(by ='car name',ascending = False,na_position='first')
```
