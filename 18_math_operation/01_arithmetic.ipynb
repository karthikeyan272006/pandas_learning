Arithmetic operation

```python
import pandas as pd
```

```python
df = pd.DataFrame([[10,20,30],
                  [5,25,15],
                  [2,20,10]])
    
df.columns =['A','B','C']
df
```

```python
# addition
# df+1
df.add(1)
```

```python
#substraction
# df - 1
# df.sub(1)
df.subtract(2)
```

```python
#multiplication
# df * 2
# df.mul(2)
# df.multiply(2)
```

```python
# division
# df /2
# df.div(2)
df.divide(2)
```

```python
# floor division or intger divison
# df // 2
df.floordiv(2)
```

```python
# modulus
# df % 2
df.mod(2)
```

```python
# power
# df ** 2
df.pow(2)
```

```python
# on series
df['A'] + 2
df['A'].divmod(2)
```
