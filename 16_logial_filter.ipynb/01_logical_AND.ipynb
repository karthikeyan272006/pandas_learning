filter the rows using AND logical operator

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/patients.csv")
df
```

```python
df['service'] == 'surgery'
```

```python
df['arrival_date'] == '2025-12-13'
```

```python
df.loc[(df['service'] == 'surgery') & (df['arrival_date'] == '2025-12-13')]
```

```python
# using three condition
df.loc[(df['age']>15) & (df['service'] == 'ICU') & (df['satisfaction']>10)]
```
