how to save the dataframe in csv

```python
import pandas as pd
```

```python
ravi =['Ravi','21','b.tech','data analyst']
rahul =['Rahul','20','b.tech','data analyst']
rekha =['Rekha','18','b.tech','data research']
karthi =['Karthi','34','b.tech','data enginnering']
professional =[ravi,rahul,rekha,karthi]
```

```python
df =pd.DataFrame(professional)
df.columns =['Name','Age',''
'degree','Role']
df

```

```python
df.to_csv('data/professional.csv')
```

```python
df.to_csv('data/professional1.csv',index=False)
```

```python
#save as semi_colon
df.to_csv('data/professional2.csv',index=False,sep=';')
```

how to save the dataframe in excel format

```python
df.to_excel('data/professional.xlsx')
```

```python
#turn of index
df.to_excel('data/professional.xlsx',index =False)
```

how to set default value for null values 
df.to_excel('data/professional.xlsx',index =False,na_rep='unknown')

```python
df.to_excel('data/professional_column.xlsx',index =False,columns=["Name",'Age'])
```
